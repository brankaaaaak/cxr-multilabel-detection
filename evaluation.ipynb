## Area Under the Curve

AUC je najvažnija metrika u medicinskoj dijagnostici jer govori koliko je model sposoban da razlikuje bolesnog pacijenta od zdravog. U ovom konkretnom obrađivanom problemu mnogo je bolji pokazatelj kvaliteta modela od Accuracy, pogotovo ako se u obzir uzme ogroman disbalans klasa. AUC je potvrda da model stvarno prepoznaje patologiju na snimku, a ne da samo pogađa najčešću klasu.

Sa desnog grafika možemo zaključiti koliko je model generalno "pametan", kada se uzmu u obzir sve bolesti zajedno. 

**Micro-average** (0.773) je učinak modela na nivou svake pojedinačne slike. 

**Macro-average** (0.770) je prosjek učinka po dijagnozama.

Činjenica da su ova dva učinka približna znači da model podjednako dobro radi i na čestim i na rijetkim bolestima. Što potvrđuje pozitivan učinak weighted_binary_crossentropy funkcije.
